```python
from ultralytics import YOLO


def main():
    model = YOLO("Models/best_model.pt")

    print("Model classes:")
    print(model.names)

    results = model.predict(
        source="Data/image_phone.jpeg",
        conf=0.25,
        save=True
    )

    print("\nDetection completed!")
    print(f"Number of result images: {len(results)}")


if __name__ == "__main__":
    main()
```
